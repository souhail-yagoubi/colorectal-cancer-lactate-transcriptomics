```python
# ======================================================================
# PROJECT 1 — COLORECTAL CANCER TRANSCRIPTOMICS
# Transcriptomic response to lactate in SW480 colorectal cancer cells
#
# Dataset: GSE180948
# Comparison: SW480 Lactate vs SW480 Control
#
# COMPLETE WORKFLOW:
#
# 1. Load raw RNA-seq counts
# 2. Check data quality
# 3. Select SW480 samples
# 4. Create metadata
# 5. Check library sizes
# 6. Filter low-expression genes
# 7. Prepare PyDESeq2 input
# 8. Normalize counts
# 9. Variance Stabilizing Transformation (VST)
# 10. PCA
# 11. Differential expression analysis
# 12. Identify significant genes
# 13. Volcano plot
# 14. Heatmap
# 15. GO / KEGG / Reactome enrichment
# 16. Save all results
# ======================================================================


# ======================================================================
# STEP 1 — IMPORT LIBRARIES
# ======================================================================

from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA

from pydeseq2.dds import DeseqDataSet
from pydeseq2.ds import DeseqStats

import gseapy as gp


# ======================================================================
# STEP 2 — DEFINE PROJECT PATHS
# ======================================================================

# Find the project root directory.
# This works whether the notebook runs from:
# - the project root
# - or the notebooks/ folder

current_dir = Path.cwd()

if current_dir.name == "notebooks":
    project_root = current_dir.parent
else:
    project_root = current_dir


data_dir = project_root / "data"
results_dir = project_root / "results"

# Create results directory if necessary
results_dir.mkdir(exist_ok=True)


print("Project root:")
print(project_root)

print("\nResults directory:")
print(results_dir)


# ======================================================================
# STEP 3 — LOAD THE RNA-SEQ COUNT MATRIX
# ======================================================================

file_path = data_dir / "GSE180948_All.counts.txt"


df = pd.read_csv(
    file_path,
    sep="\t"
)


print("\n======================================")
print("RAW DATA")
print("======================================")

print("Dataset dimensions:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())


# ======================================================================
# STEP 4 — BASIC DATA QUALITY CHECK
# ======================================================================

# All columns except AccID are RNA-seq samples

sample_cols = df.columns[1:]


print("\n======================================")
print("DATA QUALITY CHECK")
print("======================================")


# Number of samples
print("\nNumber of samples:")
print(len(sample_cols))


# Sample names
print("\nSample names:")
print(sample_cols.tolist())


# Data types
print("\nColumn data types:")
print(df[sample_cols].dtypes)


# Minimum expression count
print("\nMinimum count:")
print(df[sample_cols].min().min())


# Maximum expression count
print("\nMaximum count:")
print(df[sample_cols].max().max())


# Missing expression values
print("\nMissing expression values:")
print(
    df[sample_cols]
    .isna()
    .sum()
    .sum()
)


# Check if all counts are integers
all_integer = (
    (df[sample_cols] % 1 == 0)
    .all()
    .all()
)

print("\nAll expression values are integers:")
print(all_integer)


# ======================================================================
# STEP 5 — CHECK GENE IDENTIFIERS
# ======================================================================

print("\n======================================")
print("GENE IDENTIFIER CHECK")
print("======================================")


print("\nTotal number of rows:")
print(len(df))


print("\nUnique AccID:")
print(
    df["AccID"]
    .nunique()
)


print("\nDuplicated AccID:")
print(
    df["AccID"]
    .duplicated()
    .sum()
)


print("\nMissing AccID:")
print(
    df["AccID"]
    .isna()
    .sum()
)


# Genes with zero counts in every sample
zero_genes = (
    df[sample_cols]
    .sum(axis=1)
    .eq(0)
    .sum()
)


print("\nGenes with zero counts in all samples:")
print(zero_genes)


# ======================================================================
# STEP 6 — SELECT ONLY SW480 SAMPLES
# ======================================================================

# We analyse only one colorectal cancer cell line first.
#
# LA = Lactate
# NC = Control


sw480_cols = [

    "SW480_LA_1",
    "SW480_LA_2",
    "SW480_LA_3",

    "SW480_NC_1",
    "SW480_NC_2",
    "SW480_NC_3"

]


df_sw480 = df[
    ["AccID"] + sw480_cols
].copy()


print("\n======================================")
print("SW480 DATA")
print("======================================")


print("\nSW480 dimensions:")
print(df_sw480.shape)


display(
    df_sw480.head()
)


# ======================================================================
# STEP 7 — CREATE SAMPLE METADATA
# ======================================================================

# Metadata tells DESeq2 which samples belong
# to which experimental condition.


metadata = pd.DataFrame({

    "sample": sw480_cols,

    "condition": [

        "Lactate",
        "Lactate",
        "Lactate",

        "Control",
        "Control",
        "Control"

    ]

})


metadata = metadata.set_index(
    "sample"
)


print("\n======================================")
print("METADATA")
print("======================================")


display(metadata)


# ======================================================================
# STEP 8 — CHECK LIBRARY SIZES
# ======================================================================

# Library size = total number of counts
# measured for each sample.


library_sizes = (
    df_sw480[
        sw480_cols
    ]
    .sum()
)


print("\n======================================")
print("LIBRARY SIZES")
print("======================================")


display(library_sizes)


# Plot library sizes

plt.figure(
    figsize=(9, 5)
)


library_sizes.plot(
    kind="bar"
)


plt.ylabel(
    "Total counts"
)

plt.xlabel(
    "Samples"
)

plt.title(
    "Library Size — SW480 Samples"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()


plt.savefig(
    results_dir / "library_sizes.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ======================================================================
# STEP 9 — FILTER LOW-EXPRESSION GENES
# ======================================================================

# Rule:
#
# Keep a gene if it has:
#
# >= 10 counts
#
# in at least:
#
# 3 samples


counts = df_sw480[
    sw480_cols
]


keep = (
    (counts >= 10)
    .sum(axis=1)
    >= 3
)


df_filtered = df_sw480.loc[
    keep
].copy()


print("\n======================================")
print("GENE FILTERING")
print("======================================")


print(
    "Genes before filtering:",
    df_sw480.shape[0]
)


print(
    "Genes after filtering:",
    df_filtered.shape[0]
)


print(
    "Genes removed:",
    df_sw480.shape[0]
    -
    df_filtered.shape[0]
)


# ======================================================================
# STEP 10 — PREPARE COUNTS FOR PYDESEQ2
# ======================================================================

# PyDESeq2 requires:
#
# rows    = samples
# columns = genes


counts_df = (

    df_filtered

    .set_index(
        "AccID"
    )[sw480_cols]

    .T

    .astype(int)

)


print("\n======================================")
print("PYDESEQ2 INPUT")
print("======================================")


print(
    "Count matrix dimensions:"
)

print(
    counts_df.shape
)


display(
    counts_df.iloc[
        :5,
        :5
    ]
)


# ======================================================================
# STEP 11 — CHECK SAMPLE ORDER
# ======================================================================

print("\nSamples in count matrix:")

print(
    counts_df.index.tolist()
)


print("\nSamples in metadata:")

print(
    metadata.index.tolist()
)


same_order = (
    counts_df.index
    .equals(
        metadata.index
    )
)


print("\nSame sample order:")

print(
    same_order
)


if not same_order:
    raise ValueError(
        "Counts and metadata samples are not aligned."
    )


# ======================================================================
# STEP 12 — CREATE DESEQ2 DATASET
# ======================================================================

dds = DeseqDataSet(

    counts=counts_df,

    metadata=metadata,

    design="~condition",

    refit_cooks=True

)


print("\nDESeq2 dataset created.")


# ======================================================================
# STEP 13 — NORMALIZATION
# ======================================================================

# DESeq2 estimates size factors.
#
# These correct differences in sequencing depth
# between samples.


dds.fit_size_factors()


print("\n======================================")
print("NORMALIZATION")
print("======================================")


print(
    "Size factors:"
)

print(
    dds.obs[
        "size_factors"
    ]
)


# ======================================================================
# STEP 14 — VARIANCE STABILIZING TRANSFORMATION
# ======================================================================

# VST is used for:
#
# - PCA
# - clustering
# - visualization
#
# VST is NOT used for the differential expression test.


dds.vst(
    use_design=False
)


vst_counts = (
    dds.layers[
        "vst_counts"
    ]
)


print("\nVST matrix dimensions:")

print(
    vst_counts.shape
)


# ======================================================================
# STEP 15 — PRINCIPAL COMPONENT ANALYSIS
# ======================================================================

pca = PCA(
    n_components=2
)


pca_result = (
    pca.fit_transform(
        vst_counts
    )
)


pca_df = pd.DataFrame(

    pca_result,

    index=counts_df.index,

    columns=[
        "PC1",
        "PC2"
    ]

)


pca_df[
    "condition"
] = metadata[
    "condition"
]


print("\n======================================")
print("PCA RESULTS")
print("======================================")


display(
    pca_df
)


# Explained variance

pc1_var = (
    pca.explained_variance_ratio_[0]
    * 100
)


pc2_var = (
    pca.explained_variance_ratio_[1]
    * 100
)


print(
    f"PC1 explained variance: "
    f"{pc1_var:.1f}%"
)


print(
    f"PC2 explained variance: "
    f"{pc2_var:.1f}%"
)


print(
    f"PC1 + PC2: "
    f"{pc1_var + pc2_var:.1f}%"
)


# ======================================================================
# STEP 16 — PCA PLOT
# ======================================================================

plt.figure(
    figsize=(8, 6)
)


for condition in [

    "Control",
    "Lactate"

]:

    subset = pca_df[
        pca_df[
            "condition"
        ]
        ==
        condition
    ]


    plt.scatter(

        subset["PC1"],

        subset["PC2"],

        s=100,

        label=condition

    )


# Add sample names

for sample in pca_df.index:

    plt.annotate(

        sample,

        (

            pca_df.loc[
                sample,
                "PC1"
            ],

            pca_df.loc[
                sample,
                "PC2"
            ]

        ),

        xytext=(5, 5),

        textcoords=
        "offset points"

    )


plt.xlabel(
    f"PC1 ({pc1_var:.1f}%)"
)


plt.ylabel(
    f"PC2 ({pc2_var:.1f}%)"
)


plt.title(
    "PCA — SW480 Control vs Lactate"
)


plt.legend()


plt.tight_layout()


# IMPORTANT:
# Save before plt.show()

plt.savefig(

    results_dir / "PCA.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ======================================================================
# STEP 17 — RUN DIFFERENTIAL EXPRESSION ANALYSIS
# ======================================================================

# DESeq2 uses the RAW counts internally.
#
# We compare:
#
# Lactate vs Control


dds.deseq2()


print("\nDESeq2 model completed.")


# ======================================================================
# STEP 18 — DEFINE CONTRAST
# ======================================================================

stat_res = DeseqStats(

    dds,

    contrast=[

        "condition",

        "Lactate",

        "Control"

    ]

)


stat_res.summary()


# ======================================================================
# STEP 19 — EXTRACT DESEQ2 RESULTS
# ======================================================================

results = (
    stat_res
    .results_df
    .copy()
)


print("\n======================================")
print("DESEQ2 RESULTS")
print("======================================")


print(
    "Genes analysed:",
    len(results)
)


display(
    results.head()
)


# Important columns:
#
# baseMean
#   average normalized expression
#
# log2FoldChange
#   > 0 = higher in Lactate
#   < 0 = lower in Lactate
#
# pvalue
#   raw statistical p-value
#
# padj
#   multiple-testing corrected p-value


# ======================================================================
# STEP 20 — SORT MOST SIGNIFICANT GENES
# ======================================================================

results_sorted = (
    results
    .sort_values(
        "padj"
    )
)


print(
    "\nTop 20 most significant genes:"
)


display(

    results_sorted[

        [
            "baseMean",
            "log2FoldChange",
            "pvalue",
            "padj"
        ]

    ].head(20)

)


# ======================================================================
# STEP 21 — DEFINE SIGNIFICANT GENES
# ======================================================================

# Significance criteria:
#
# adjusted p-value < 0.05
#
# AND
#
# |log2FoldChange| >= 1
#
# log2FC = 1 roughly corresponds to a 2-fold increase.
# log2FC = -1 roughly corresponds to a 2-fold decrease.


significant = results[

    (
        results[
            "padj"
        ]
        < 0.05
    )

    &

    (
        results[
            "log2FoldChange"
        ].abs()
        >= 1
    )

].copy()


# ======================================================================
# STEP 22 — UPREGULATED GENES
# ======================================================================

upregulated = significant[

    significant[
        "log2FoldChange"
    ]
    >= 1

].copy()


# ======================================================================
# STEP 23 — DOWNREGULATED GENES
# ======================================================================

downregulated = significant[

    significant[
        "log2FoldChange"
    ]
    <= -1

].copy()


# ======================================================================
# STEP 24 — DIFFERENTIAL EXPRESSION SUMMARY
# ======================================================================

print("\n======================================")
print("DIFFERENTIAL EXPRESSION SUMMARY")
print("======================================")


print(
    "Genes analysed:",
    len(results)
)


print(
    "Significant genes:",
    len(significant)
)


print(
    "Upregulated:",
    len(upregulated)
)


print(
    "Downregulated:",
    len(downregulated)
)


# ======================================================================
# STEP 25 — TOP UPREGULATED GENES
# ======================================================================

print(
    "\nTOP 10 UPREGULATED GENES"
)


display(

    upregulated

    .sort_values(
        "padj"
    )[

        [
            "baseMean",
            "log2FoldChange",
            "padj"
        ]

    ]

    .head(10)

)


# ======================================================================
# STEP 26 — TOP DOWNREGULATED GENES
# ======================================================================

print(
    "\nTOP 10 DOWNREGULATED GENES"
)


display(

    downregulated

    .sort_values(
        "padj"
    )[

        [
            "baseMean",
            "log2FoldChange",
            "padj"
        ]

    ]

    .head(10)

)


# ======================================================================
# STEP 27 — SAVE DESEQ2 TABLES
# ======================================================================

results.to_csv(
    results_dir /
    "deseq2_all_results.csv"
)


significant.to_csv(
    results_dir /
    "deseq2_significant_genes.csv"
)


upregulated.to_csv(
    results_dir /
    "upregulated_genes.csv"
)


downregulated.to_csv(
    results_dir /
    "downregulated_genes.csv"
)


print(
    "\nDESeq2 result tables saved."
)


# ======================================================================
# STEP 28 — PREPARE VOLCANO PLOT
# ======================================================================

volcano = results.copy()


# Remove genes without usable values

volcano = volcano.dropna(

    subset=[

        "padj",

        "log2FoldChange"

    ]

)


# Calculate -log10 adjusted p-value

volcano[
    "minus_log10_padj"
] = -np.log10(

    volcano[
        "padj"
    ].clip(
        lower=1e-300
    )

)


# Default category

volcano[
    "status"
] = "Not significant"


# Upregulated genes

volcano.loc[

    (
        volcano[
            "padj"
        ]
        < 0.05
    )

    &

    (
        volcano[
            "log2FoldChange"
        ]
        >= 1
    ),

    "status"

] = "Upregulated"


# Downregulated genes

volcano.loc[

    (
        volcano[
            "padj"
        ]
        < 0.05
    )

    &

    (
        volcano[
            "log2FoldChange"
        ]
        <= -1
    ),

    "status"

] = "Downregulated"


# ======================================================================
# STEP 29 — VOLCANO PLOT
# ======================================================================

plt.figure(
    figsize=(9, 6)
)


for status in [

    "Not significant",

    "Upregulated",

    "Downregulated"

]:

    subset = volcano[
        volcano[
            "status"
        ]
        ==
        status
    ]


    plt.scatter(

        subset[
            "log2FoldChange"
        ],

        subset[
            "minus_log10_padj"
        ],

        s=15,

        alpha=0.6,

        label=status

    )


# Fold-change thresholds

plt.axvline(
    x=1,
    linestyle="--"
)


plt.axvline(
    x=-1,
    linestyle="--"
)


# Statistical significance threshold

plt.axhline(

    y=-np.log10(
        0.05
    ),

    linestyle="--"

)


plt.xlabel(
    "log2 Fold Change"
)


plt.ylabel(
    "-log10 adjusted p-value"
)


plt.title(
    "SW480 — Lactate vs Control"
)


plt.legend()


plt.tight_layout()


plt.savefig(

    results_dir /
    "volcano_plot.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ======================================================================
# STEP 30 — SELECT TOP 20 GENES FOR HEATMAP
# ======================================================================

top_genes = (

    significant

    .sort_values(
        "padj"
    )

    .head(20)

    .index

    .tolist()

)


print(
    "\nTop 20 genes selected for heatmap:"
)


print(
    top_genes
)


# ======================================================================
# STEP 31 — CREATE VST DATAFRAME
# ======================================================================

vst_df = pd.DataFrame(

    vst_counts,

    index=counts_df.index,

    columns=counts_df.columns

)


# Keep only top genes

heatmap_data = (

    vst_df[
        top_genes
    ]

    .T

)


print(
    "\nHeatmap dimensions:"
)


print(
    heatmap_data.shape
)


# ======================================================================
# STEP 32 — STANDARDIZE EACH GENE
# ======================================================================

# Z-score by gene.
#
# Positive value:
# higher than the gene's average expression.
#
# Negative value:
# lower than the gene's average expression.


heatmap_z = heatmap_data.sub(

    heatmap_data.mean(
        axis=1
    ),

    axis=0

)


heatmap_z = heatmap_z.div(

    heatmap_data.std(
        axis=1
    ).replace(
        0,
        1
    ),

    axis=0

)


# ======================================================================
# STEP 33 — HEATMAP
# ======================================================================

plt.figure(
    figsize=(9, 10)
)


image = plt.imshow(

    heatmap_z,

    aspect="auto"

)


plt.colorbar(

    image,

    label="Z-score expression"

)


plt.xticks(

    range(
        len(
            heatmap_z.columns
        )
    ),

    heatmap_z.columns,

    rotation=45,

    ha="right"

)


plt.yticks(

    range(
        len(
            heatmap_z.index
        )
    ),

    heatmap_z.index

)


plt.xlabel(
    "Samples"
)


plt.ylabel(
    "Genes"
)


plt.title(
    "Top 20 Differentially Expressed Genes"
)


plt.tight_layout()


plt.savefig(

    results_dir /
    "top20_heatmap.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ======================================================================
# STEP 34 — PREPARE GENE LISTS FOR FUNCTIONAL ENRICHMENT
# ======================================================================

up_genes = (

    upregulated

    .index

    .dropna()

    .astype(str)

    .tolist()

)


down_genes = (

    downregulated

    .index

    .dropna()

    .astype(str)

    .tolist()

)


print(
    "\nUpregulated genes:",
    len(up_genes)
)


print(
    "Downregulated genes:",
    len(down_genes)
)


# ======================================================================
# STEP 35 — DEFINE ENRICHMENT DATABASES
# ======================================================================

gene_sets = [

    "GO_Biological_Process_2023",

    "KEGG_2021_Human",

    "Reactome_2022"

]


# ======================================================================
# STEP 36 — ENRICHMENT OF UPREGULATED GENES
# ======================================================================

enr_up = gp.enrichr(

    gene_list=up_genes,

    gene_sets=gene_sets,

    organism="human",

    outdir=None

)


up_enrichment = (
    enr_up
    .results
    .copy()
)


# ======================================================================
# STEP 37 — ENRICHMENT OF DOWNREGULATED GENES
# ======================================================================

enr_down = gp.enrichr(

    gene_list=down_genes,

    gene_sets=gene_sets,

    organism="human",

    outdir=None

)


down_enrichment = (
    enr_down
    .results
    .copy()
)


# ======================================================================
# STEP 38 — KEEP SIGNIFICANT PATHWAYS
# ======================================================================

sig_up_pathways = (

    up_enrichment[

        up_enrichment[
            "Adjusted P-value"
        ]
        < 0.05

    ]

    .copy()

)


sig_down_pathways = (

    down_enrichment[

        down_enrichment[
            "Adjusted P-value"
        ]
        < 0.05

    ]

    .copy()

)


print("\n======================================")
print("FUNCTIONAL ENRICHMENT")
print("======================================")


print(
    "Significant UP pathways:",
    len(sig_up_pathways)
)


print(
    "Significant DOWN pathways:",
    len(sig_down_pathways)
)


# ======================================================================
# STEP 39 — DISPLAY TOP ENRICHED PATHWAYS
# ======================================================================

print(
    "\nTOP 15 UPREGULATED PATHWAYS"
)


display(

    sig_up_pathways

    .sort_values(
        "Adjusted P-value"
    )[

        [
            "Gene_set",
            "Term",
            "Adjusted P-value",
            "Combined Score",
            "Genes"
        ]

    ]

    .head(15)

)


print(
    "\nTOP 15 DOWNREGULATED PATHWAYS"
)


display(

    sig_down_pathways

    .sort_values(
        "Adjusted P-value"
    )[

        [
            "Gene_set",
            "Term",
            "Adjusted P-value",
            "Combined Score",
            "Genes"
        ]

    ]

    .head(15)

)


# ======================================================================
# STEP 40 — SAVE ENRICHMENT TABLES
# ======================================================================

up_enrichment.to_csv(

    results_dir /
    "enrichment_upregulated.csv",

    index=False

)


down_enrichment.to_csv(

    results_dir /
    "enrichment_downregulated.csv",

    index=False

)


# ======================================================================
# STEP 41 — PLOT TOP ENRICHED UPREGULATED PATHWAYS
# ======================================================================

if len(sig_up_pathways) > 0:

    plot_data = (

        sig_up_pathways

        .sort_values(
            "Adjusted P-value"
        )

        .head(15)

        .copy()

    )


    plot_data[
        "minus_log10_padj"
    ] = -np.log10(

        plot_data[
            "Adjusted P-value"
        ]

    )


    plot_data = (

        plot_data

        .sort_values(
            "minus_log10_padj"
        )

    )


    plt.figure(
        figsize=(10, 8)
    )


    plt.barh(

        plot_data[
            "Term"
        ],

        plot_data[
            "minus_log10_padj"
        ]

    )


    plt.xlabel(
        "-log10 Adjusted P-value"
    )


    plt.ylabel(
        "Biological pathway"
    )


    plt.title(

        "Functional Enrichment of Upregulated Genes\n"
        "SW480 Lactate vs Control"

    )


    plt.tight_layout()


    plt.savefig(

        results_dir /
        "top_enriched_pathways.png",

        dpi=300,

        bbox_inches="tight"

    )


    plt.show()


# ======================================================================
# STEP 42 — FINAL PROJECT SUMMARY
# ======================================================================

print("\n")
print("======================================================")
print("PROJECT 1 — FINAL SUMMARY")
print("======================================================")


print(
    "Cell line:",
    "SW480"
)


print(
    "Comparison:",
    "Lactate vs Control"
)


print(
    "\nGenes analysed:",
    len(results)
)


print(
    "Significant genes:",
    len(significant)
)


print(
    "Upregulated genes:",
    len(upregulated)
)


print(
    "Downregulated genes:",
    len(downregulated)
)


print(
    "\nPCA:"
)


print(
    f"PC1 = {pc1_var:.1f}%"
)


print(
    f"PC2 = {pc2_var:.1f}%"
)


print(
    f"PC1 + PC2 = "
    f"{pc1_var + pc2_var:.1f}%"
)


print(
    "\nSignificant UP pathways:",
    len(sig_up_pathways)
)


print(
    "Significant DOWN pathways:",
    len(sig_down_pathways)
)


print(
    "\nAll results saved in:"
)


print(
    results_dir
)


print("\n")
print("======================================================")
print("ANALYSIS COMPLETED")
print("======================================================")
```
